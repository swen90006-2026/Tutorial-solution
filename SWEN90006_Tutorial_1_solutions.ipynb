# SWEN90006 Tutorial 1 solutions

## Task 1: find_last


1. The required data type of input is a pair of (int[], int).
2. The fault is that the line  ```index = index - 2``` should be ```index = index - 1```.
The faulty version steps backwards two positions at a time, so starting from the last element it only checks every second element and never looks at the ones in between
3. Any case in which the decrement is never reached. Either the list is empty, or the last element is itself a match, so the function returns before the fault executes
4. Any case in which none of the skipped elements equals ```target_value```, or ```target_value``` is not in the non-empty list.
5. The same test cases as in task 4.
An error is created whenever the faulty statement is reached. Every execution of ```index = index - 2``` leaves ```index``` holding a value different from the one the correct program would hold at that point.
6. Replace ```index = index - 2``` with ```index = index - 1```


## Task 2: last_zero
1. The required data type of input is a list of integers int[].
2. The loop should iterate backwards instead of forwards.
3. All inputs execute the fault.
4. Any list with no zeros, or any list with exactly one zero.
5. Any non-empty list with no zeros, or any list with exactly one zero. An error is created as soon as the loop is entered when it start with ```index == 0``` where ```index``` should be index of the last element in the list.
6. Make the loop iterate backwards.

## Task 3: count_positive
1. The required data type of input is a list of integers int[].
2. ```L[index] >= 0``` should be ```L[index] > 0```: it should count only strictly positive elements
3. An empty list (```L = []```).
4. Any list with no occurrence of 0.
5. None! An error is created when the list contains a 0, ```count``` exceeds the number of number of _strictly_ positive elements
6. Replace ```L[index] >= 0``` with ```L[index] > 0```.

## Discussion

1. Through testing, it's impossible to prove that software is free of failures, instead we aim to reveal as many faults as possible.

2. To illustrate the difference between testing and debugging. In SWEN90006, we cover testing, but not debugging. As part of this exercise, we did debugging to determine where the fault is.

3. To clarify the terminology of faults, failures, and errors. Faults and failures are quite often confused, and importantly, the term "bugs" is used to mean both. In SWEN90006, the term "bugs" is disallowed due to its ambiguity.

4. To illustrate an important point about coverage-based testing (which we will cover later in the subject): just executing every statement of a program is a *necessary but not sufficient* condition for a good test suite. It is not only possible to execute a statement but not trigger an error, but to trigger an error that does NOT result in a failure.

  These cases are known as *coincidental correctness*, and are exceedingly common in real-world software.
